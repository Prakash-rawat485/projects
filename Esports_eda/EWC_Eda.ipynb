## EWC and 2025 Esports World Cup – Deep Overview for Your EDA

The **Esports World Cup (EWC)** is the world’s largest annual multi‑title esports festival by prize pool and scope, held in Riyadh, Saudi Arabia. The **2025 edition** ran from **8 July to 24 August 2025** at **Boulevard City**, featuring **26 events across 25 games** and a record **US\$71.5 million** total prize pool, crowned by **Team Falcons** as Club Champion again.[^1_1][^1_2]

Below is a structured, EDA‑friendly breakdown you can adapt into notebooks, dashboards, or reports.

***

## 1) What is the Esports World Cup (EWC)?

- **Organizer:** Esports World Cup Foundation (EWCF), a nonprofit funded by Saudi Arabia’s Public Investment Fund.[^1_2][^1_1]
- **Origin:** Successor to **Gamers8** (2022–2023) and **Gamers Without Borders** (2020–2021), rebranded and scaled up in 2024.[^1_2]
- **Core idea:** A **multi‑game, multi‑week** festival where top organizations (“clubs”) compete across many titles; winners in each game get a big prize, and the best overall club wins the **Club Championship** with a massive share of the pool.[^1_1][^1_2]
- **Location (2024–2025):** Riyadh, Saudi Arabia – Boulevard City.[^1_1][^1_2]
- **Scale:** In 2025, organizers reported **2,500+ participants** from **100+ countries**, with a **\$71.5M** prize pool.[^1_2][^1_1]

Reference:

- Esports World Cup (general) – Wikipedia: https://en.wikipedia.org/wiki/Esports_World_Cup[^1_2]

***

## 2) 2025 EWC – Key Facts at a Glance

- **Dates:** 8 July – 24 August 2025 (7 weeks).[^1_1]
- **Venue:** Boulevard City, Riyadh, Saudi Arabia.[^1_1]
- **Events:** 26 tournaments across **25 esports**.[^1_1]
- **Total prize pool:** **US\$71.5 million** (largest combined in esports history at the time).[^1_1]
- **Club Champion:** **Team Falcons** (defended 2024 title).[^1_1]
- **Global Ambassador:** Cristiano Ronaldo; **Chess Ambassador:** Magnus Carlsen.[^1_1]

Reference:

- 2025 Esports World Cup – Wikipedia: https://en.wikipedia.org/wiki/2025_Esports_World_Cup[^1_1]

This is great for EDA: you can compare **title count**, **genre mix**, and **prize allocations** year‑over‑year.

## 4) Prize Pool Structure (2025)

Total: **US\$71.5M**, split into four main buckets:[^1_1]

- **Club Championship:** \$27,000,000 (top 24 clubs by cross‑game performance)
- **Game Championships:** \$38,000,000 (pools for each title’s tournament)
- **Qualifiers:** \$6,000,000 (regional / last‑chance qualifiers, etc.)
- **MVP Awards:** \$500,000 (individual standout players)

This structure is crucial for EDA because:

- You can model **club revenue** as a function of placements across multiple games.
- You can analyze **per‑game prize concentration** (e.g., how much of the \$38M went to top 5 titles).
- You can test correlations between **club partner status** and **total earnings**.

***

## 5) Club Championship – Cross‑Game Scoring System

The Club Championship ranks “clubs” (organizations) using points from individual game tournaments.

### Qualification rules (2025)

- A club must finish **top‑8 in at least two competitions** to be eligible.
- To **win** the Club Championship, a club must also **win at least one game tournament**.[^1_1]

## 10) Controversies \& Ethical Context (Important for Narrative EDA)

Any serious analysis of EWC should acknowledge:

- **Sportswashing concerns:** Heavy criticism over using esports to soften Saudi Arabia’s human rights image.[^1_2][^1_1]
- **Player boycotts:**
    - Some players (e.g., Christopher “ChrisCCH” Hancock in SF6) refused to participate due to ethical concerns.[^1_2][^1_1]
    - Hambino (Apex Legends) declined to attend, citing LGBTQ+ safety concerns in Saudi Arabia.[^1_1]
- **GeoGuessr withdrawal:** Community map creators protested, making maps unplayable; GeoGuessr withdrew from EWC 2025.[^1_2][^1_1]

You can incorporate this into:

- Sentiment analysis of social media around EWC.
- Comparative analysis of participation rates vs controversy intensity.
- Qualitative annotation of events (e.g., flagging boycotted tournaments).

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
````

````python
summary = pd.DataFrame({
    'Data Type': df.dtypes,
    'Missing Values': df.isnull().sum(),
    'Unique Values': df.nunique()
})

print(summary)
````

````python
# Count Events
event_counts = df['Event'].value_counts()

# Plot Style
plt.figure(figsize=(14,7))
ax = sns.countplot(
    data=df,
    y='Event',
    order=event_counts.index,
    palette='viridis'
)

# Add Count and Percentage
total = len(df)

for p in ax.patches:
    count = int(p.get_width())
    percentage = 100 * count / total

    ax.text(
        p.get_width()+0.5,
        p.get_y()+p.get_height()/2,
        f'{count} ({percentage:.1f}%)',
        va='center',
        fontsize=11,
        fontweight='bold'
    )

# Title
plt.title(
    'Distribution of Medalists Across Esports Events',
    fontsize=18,
    weight='bold'
)

plt.xlabel('Number of Medalists', fontsize=13)
plt.ylabel('Esports Event', fontsize=13)

# Remove top and right borders
sns.despine()

plt.tight_layout()
plt.show()
````

**Finding**

* **Teamfight Tactics** and **Overwatch 2** have the highest number of medalists (16 each, **6.2%**), followed closely by **Counter-Strike 2, Dota 2, League of Legends,** and several other titles with **15 medalists (5.8%)** each.
* Traditional esports such as **Chess, StarCraft II, Street Fighter 6, Tekken 8,** and **EA Sports FC 25** have the fewest medalists (**3 each, 1.2%**), indicating that medal distribution is concentrated in a few popular events.


````python
# Medal Counts
medal_counts = df['Medal'].value_counts()

# Figure
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# -------------------------
# Countplot
# -------------------------
ax = sns.countplot(
    data=df,
    x='Medal',
    order=medal_counts.index,
    palette=['gold', 'silver', '#CD7F32'],
    ax=axes[0]
)

total = len(df)

for p in ax.patches:
    count = int(p.get_height())
    percentage = count / total * 100

    ax.annotate(
        f'{count}\n({percentage:.1f}%)',
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha='center',
        va='bottom',
        fontsize=11,
        fontweight='bold'
    )

axes[0].set_title("Medal Count Distribution", fontsize=15, fontweight='bold')
axes[0].set_xlabel("Medal")
axes[0].set_ylabel("Number of Medalists")

# -------------------------
# Pie Chart
# -------------------------
axes[1].pie(
    medal_counts,
    labels=medal_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['gold', 'silver', '#CD7F32'],
    explode=[0.05, 0.05, 0.05],
    shadow=True,
    textprops={'fontsize':11, 'fontweight':'bold'}
)

axes[1].set_title("Percentage Distribution of Medals",
                  fontsize=15,
                  fontweight='bold')

# Main Title
plt.suptitle(
    "Medal Distribution Analysis",
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
````

### **Findings**

* The medal distribution is **highly balanced**, with **Bronze medals (88, 34.2%)** slightly exceeding **Gold (85, 33.1%)** and **Silver (84, 32.7%)** medals.
* The small difference among medal counts indicates a **fair and evenly distributed competition**, with no single medal category dominating the overall results.


````python
# Top 15 Countries
country_counts = df['Country'].value_counts().head(15)

# Figure
plt.figure(figsize=(12,8))

ax = sns.barplot(
    x=country_counts.values,
    y=country_counts.index,
    palette="viridis"
)

# Add Count + Percentage
total = len(df)

for i, value in enumerate(country_counts.values):
    percentage = (value / total) * 100

    ax.text(
        value + 0.3,
        i,
        f'{value} ({percentage:.1f}%)',
        va='center',
        fontsize=11,
        fontweight='bold'
    )

# Titles
plt.title(
    "Top 15 Countries by Number of Players",
    fontsize=18,
    fontweight='bold'
)

plt.xlabel("Number of Players", fontsize=13)
plt.ylabel("Country", fontsize=13)

sns.despine()
plt.tight_layout()
plt.show()

# -------------------------
# Summary
# -------------------------

print("="*45)
print("Top Country :", country_counts.idxmax())
print("Players     :", country_counts.max())

print("\nTop 15 Countries")
print(country_counts)
````

### Findings

- The medal distribution is nearly balanced across all three categories, with **Bronze** accounting for **88 medals (34.2%)**, **Gold** for **85 medals (33.1%)**, and **Silver** for **84 medals (32.7%)**.
- Bronze medals have the highest count, but the difference compared to Gold and Silver is very small.
- The close distribution indicates that the competition was highly balanced, with no single medal category dominating the overall results.
- Overall, the medal allocation reflects a fair and competitive tournament across all esports events.

````python
# Filter only Gold medalists
gold_df = df[df["Medal"] == "Gold"]

# Count Gold medals in each Event
gold_event = (
    gold_df["Event"]
    .value_counts()
    .sort_values(ascending=True)
)

# Plot
plt.figure(figsize=(12,8))

sns.barplot(
    x=gold_event.values,
    y=gold_event.index
)

plt.title("Number of Gold Medalists by Event", fontsize=16, fontweight="bold")
plt.xlabel("Number of Gold Medalists", fontsize=12)
plt.ylabel("Event", fontsize=12)

# Add values on bars
for i, value in enumerate(gold_event.values):
    plt.text(value + 0.2, i, str(value), va='center')

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
````

### Findings

- **Overwatch 2** recorded the highest number of **Gold medalists (6)**, making it the top-performing event in terms of gold medals.
- **Counter-Strike 2, League of Legends, Dota 2, CrossFire, Honor of Kings, Rainbow Six Siege X,** and **Mobile Legends: Bang Bang (Men & Women)** each secured **5 Gold medalists**, indicating strong competition across these titles.
- Mid-performing events such as **PUBG Mobile, Free Fire, Rennsport, PUBG: Battlegrounds, Call of Duty: Black Ops 6,** and **Teamfight Tactics** recorded **4 Gold medalists** each.
- Traditional esports titles including **Chess, StarCraft II, Street Fighter 6, Tekken 8, EA Sports FC 25,** and **Fatal Fury: City of the Wolves** had the fewest Gold medalists (**1 each**), reflecting their smaller medal allocation.

````python
# Create Country vs Medal table
country_medal = pd.crosstab(df["Country"], df["Medal"])

# Select Top 15 countries by total medals
top15 = country_medal.sum(axis=1).sort_values(ascending=False).head(15).index
country_medal = country_medal.loc[top15]

# Plot
country_medal.plot(
    kind="bar",
    stacked=True,
    figsize=(14,7)
)

plt.title("Top 15 Countries by Medal Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Country")
plt.ylabel("Number of Medals")
plt.xticks(rotation=45, ha="right")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
````

### Findings

- **China** dominates the medal table with the highest total medals, followed by **South Korea** and the **Philippines**, highlighting their strong performance across multiple esports events.
- The **USA** and **Russia** also recorded impressive medal counts, while **France, Brazil,** and **Saudi Arabia** formed the mid-performing group among the top 15 countries.
- Most leading countries earned a balanced mix of **Gold, Silver,** and **Bronze** medals, indicating consistent performance rather than reliance on a single medal category.
- Countries such as **Indonesia, the UK, Turkey, Ukraine,** and **the Netherlands** secured comparatively fewer medals but still ranked among the top 15, reflecting competitive participation in the championship.

````python
# Team vs Medal table
team_medal = pd.crosstab(df["Team_Organization"], df["Medal"])

# Top 15 teams by total medals
top15 = team_medal.sum(axis=1).sort_values(ascending=False).head(15).index
team_medal = team_medal.loc[top15]

# Plot
team_medal.plot(
    kind="barh",
    stacked=True,
    figsize=(12,8)
)

plt.title("Top 15 Team Organizations by Medal Count", fontsize=16, fontweight="bold")
plt.xlabel("Number of Medals")
plt.ylabel("Team Organization")

plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.tight_layout()
plt.show()
````
